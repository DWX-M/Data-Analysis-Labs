VARIABILITY ESTIMATION

```python
import numpy as np
quiz_scores = [75, 65, 89, 55, 78]
assignment_scores = [82, 78, 90, 70, 80]
midterm_scores = [90, 72, 85, 60, 82]
project_scores = [88, 80, 92, 65, 85]
final_exam_scores = [85, 75, 88, 99, 100]
```

 Mean Absolute Deviation (MAD) for ‘Quiz_Scores’

```python
mean_quiz = np.mean(quiz_scores)
mad_quiz = np.mean(np.abs(quiz_scores - mean_quiz))
print("Mean Absolute Deviation (MAD):", mad_quiz)
```

Calculating Population Variance (𝝈𝟐) for ‘Assignment_Scores’.

```python
variance_assignment = np.var(assignment_scores, ddof=0)
print("Population Variance of Assignment Score (dof=0):",
variance_assignment)
```

3. Degree of freedom for ‘Midterm_Scores’ (Sample variance (𝒔𝟐)).

```python
# The degree of freedom is n-1, where n is the number of

df_midterm = len(midterm_scores) - 1
print("Sample Variance of Midterm Scores (dof=1):",
df_midterm)
```

 Standard deviation for ‘Project_Scores’ (Population (𝜎))

```python
pop_std_project = np.std(project_scores)
print("Standard deviation of Project_Scores (Population):",
pop_std_project)
```

# Standard deviation for ‘Project_Scores’ (Sample (𝑠)).


```python
sample_std_project = np.std(project_scores, ddof=1)
print("Standard deviation of Project_Scores (Sample):",
sample_std_project)
```

Range for ‘Final_Exam_Scores’

```python
range_final_exam = np.max(final_exam_scores) - np.min(final_exam_scores)
print("Range of final exam score:", range_final_exam)
```

Calculate the 25th, 50th and 75th percentile for ‘Quiz_Scores’.

```python
Q1 = np.percentile(quiz_scores, 25)
Q2 = np.percentile(quiz_scores, 50)
Q3 = np.percentile(quiz_scores, 75)
print(f"First Quartile (Q1): {Q1}")
print(f"Second Quartile (Q2): {Q2}")
print(f"Third Quartile (Q3): {Q3}")
```

 Inter Quartile Range (IQR) for ‘Quiz_Scores’.


```python
IQR = Q3 - Q1
print(f"Interquartile Range (IQR): {IQR}")
```

10. Crating a Boxplot of all columns.


```python
import matplotlib.pyplot as plt
# Create subplots for each boxplot
fig, axs = plt.subplots(1, 5, figsize=(20, 5))
# Boxplot for Quiz Scores
axs[0].boxplot(quiz_scores)
axs[0].set_title('Quiz Scores')
# Boxplot for Assignment Scores
axs[1].boxplot(assignment_scores)
axs[1].set_title('Assignment Scores')
# Boxplot for Midterm Scores
axs[2].boxplot(midterm_scores)
axs[2].set_title('Midterm Scores')
# Boxplot for Project Scores
axs[3].boxplot(project_scores)
axs[3].set_title('Project Scores')
# Boxplot for Final Exam Scores
axs[4].boxplot(final_exam_scores)
axs[4].set_title('Final Exam Scores')
# Display the boxplot
plt.tight_layout()
plt.show()
```
