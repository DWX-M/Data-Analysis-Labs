```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
sns.set(color_codes=True)
```

1. Upload the CSV dataset from the PC and read it into the data frame.


```python
from google.colab import files
# Use the file upload dialog to upload your CSV file
uploaded = files.upload()
# Read the CSV file into a data frame.
df = pd.read_csv('data.csv')
#print(df)
df
```

2. Display the First and the Last 5 rows from the data frame.

```python
# To display the top 5 rows
df.head(5)
# To display the bottom 5 rows
df.tail(5)
#Checking the types of data
df.dtypes
```

3. Dropping irrelevant columns.

```python
df = df.drop(['Engine Fuel Type', 'Market Category', 'Vehicle Style', 'Popularity', 'Number of Doors', 'Vehicle Size'],
axis=1)
df.head(5)
```

```python
# Renaming the columns
df = df.rename(columns={"Engine HP": "HP", "Engine Cylinders": "Cylinders", "Transmission Type": "Transmission", "Driven_Wheels": "Drive Mode","highway MPG": "MPG-H", "city mpg": "MPG-C", "MSRP": "Price" })
df.head(5)
```

```python
# Accessing rows and columns of the DataFrame
rows, columns = df.shape
print(f"The DataFrame has {rows} rows and {columns} columns.")
```

4. Showing duplicate rows

```python
# Keep='first' marks duplicates as True except for the first occurrence
duplicates = df[df.duplicated(keep='first')]
#print(duplicates)
duplicates
```

```python
#Counting the duplicate rows
duplicate_rows_df = df[df.duplicated()]
print("Number of duplicate rows: ", duplicate_rows_df.shape)
# Count the number of rows before removing duplicate
df.count()
```

5. Dropping duplicate records

```python
df = df.drop_duplicates()
#Display data after duplicate removal
df.head(5)
```

```python
# Count the number of rows after removing duplicate
df.count()
```

6. Plot a Bar Chart for Car-making companies.

```python
df.Make.value_counts().nlargest(40).plot(kind='bar',
figsize=(10,5))
plt.title("Number of cars by make")
plt.ylabel('Number of cars')
plt.xlabel('Make');
```

7. Plot Histogram for the numerical feature (HP).

```python
plt.hist(df['HP'], bins=10, color='steelblue', edgecolor='black')
plt.title('Distribution of HP')
plt.xlabel('HP')
plt.ylabel('Frequency')
plt.show()
```

8. Scatter Plot between Horsepower and Price. :


```python
fig, ax = plt.subplots(figsize=(10,6))
ax.scatter(df['HP'], df['Price'])
ax.set_xlabel('HP')
ax.set_ylabel('Price')
plt.show()
```

9. Draw a heatmap to find the relationship between independent and dependent
variables.

```python
plt.figure(figsize=(10, 5))
c = df.corr(numeric_only=True)
sns.heatmap(c, cmap="BrBG", annot=True)
plt.show()
```
