```python
#Importing dependencies
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
style.use('ggplot') # 'ggplot' style is one of the predefined styles available in matplotlib.
```

```python
# Creating array list
x = np.array([1, 2, 3, 4, 5, 6])
y=np.array([6,7,8,9,10,11])
```

```python
variance = np.var(x)
# Print the variance
print("Variance:", variance)
```

```python
# Calculate the covariance
np.sum((x- x.mean()) * (y - y.mean())) / (x.shape[0] - 1)
```

```python
plt.axvline(x.mean(),c='g',label='mean-x')
plt.axhline(y.mean(),c='y',label='mean-y')
plt.scatter(x,y,label='x-y plot')
plt.legend()
```

```python
# Zero Covariance
x = np.array([1, 1, 1, 1, 1, 1]) # X is stable
y = np.array([5, 6, 7, 8, 9, 10]) # Y increasing
# Calculate the covariance matrix
np.cov(x), np.cov(y), np.cov(x,y)
```

```python
# Plotting (Zero)covariance graph
plt.axvline(x.mean(),c='g',label='mean-x')
plt.axhline(y.mean(),c='y',label='mean-y')
plt.scatter(x,y,label='x-y plot')
plt.legend()
```

```python
# Negative Covariance
x = np.array([1, 2, 3, 4, 5, 6]) # Increasing values of x
y = np.array([10, 9, 8, 7, 6, 5]) # Decreasing values of y
#Calculate the covariance between x and y
np.cov(x), np.cov(y), np.cov(x,y)
```

```python
#Ploting covariance graph (-Ve)
plt.axvline(x.mean(),c='g',label='mean-x')
plt.axhline(y.mean(),c='y',label='mean-y')
plt.scatter(x,y,label='x-y plot')
plt.legend()
```

CORRELATION
Correlation is a statistical measure that quantifies the degree to which two or more variables
are related or move together in a linear fashion. It assesses the strength and direction of the
linear relationship between variables.

```python
# Creating array list
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([15, 16, 17, 18, 19, 20])
```

```python
#Calculate the covariance between x and y
cov_mat = np.cov(x,y)
print(cov_mat)
```

```python
# 4. PEARSON CORRELATION
cov_mat[1][0] / np.sqrt(cov_mat[0][0] * cov_mat[1][1])
```

```python
#Ploting correlation graph (Perfect +Ve Correlation)
plt.axvline(x.mean(),c='g',label='mean-x')
plt.axhline(y.mean(),c='y',label='mean-y')
plt.scatter(x,y,label='x-y plot')
plt.legend()
```

```python
# ALTERNATE METHOD FOR PEARSON CORRELATION
np.corrcoef(x,y)
```

```python
from scipy.stats import pearsonr
pearsonr(x,y)
```

```python
# Negative Correlation (x increasing and y decreasing)
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([15, 16, 17, 18, 19, 20][::-1])
```

```python
# Plotting negative correlation graph
plt.axvline(x.mean(),c='g',label='mean-x')
plt.axhline(y.mean(),c='y',label='mean-y')
plt.scatter(x,y,label='x-y plot')
plt.legend()
```

```python
# Calculating correlation
np.corrcoef(x,y)
```

```python
# Calculating Pearson(r) correlation coefficient
pearsonr(x,y)
```

```python
# ZERO Correlation
x = np.array([1, 1, 1, 1, 1, 1])
y = np.array([15, 16, 17, 18, 19, 20])
# Covariance Matrix
cov_mat = np.cov(x,y)
print(cov_mat)
```

```python
#PEARSON CORRELATION
cov_mat[1][0] / np.sqrt(cov_mat[0][0] * cov_mat[1][1])
"""Note: If either cov_mat[0][0] or cov_mat[1][1] is equal to zero, you will encounter division by zero when calculating the correlation coefficient, leading to a nan result. """
```
